```python
hari = "Sabtu"

if hari == "Sabtu" or hari == "Minggu":
    diskon = 0.20  # Diskon 20%
    print(f"Selamat! Anda mendapat diskon akhir pekan sebesar {diskon*100}%.")
elif hari == "Jumat":
    diskon = 0.15  # Diskon 15%
    print(f"Diskon spesial Jumat! Anda mendapat potongan {diskon*100}%.")
else:
    diskon = 0.05  # Diskon 5%
    print(f"Anda mendapat diskon hari biasa sebesar {diskon*100}%.")
```

```python
#Latihan
#Nama:Danu Marsya
#Nim:4232501041
#Soal 2.Menentukan Tahun Kabisat dengan menggunakan kondisi a. Penjelasan:
#- Syarat tahun kabisat:Habis dibagi 400 → kabisat.Habis dibagi 4 dan tidak habis dibagi 100 → kabisat.

tahun = int(input("Masukkan tahun: "))

if tahun % 400 == 0:
    print("Tahun kabisat")
elif tahun % 4 == 0 and tahun % 100 != 0:
    print("Tahun kabisat")
else:
    print("Bukan tahun kabisat")
```
