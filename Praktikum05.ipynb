```python
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_satuan = int(input("Masukkan harga per buku (Rp): "))
uang_bayar = int(input("Masukkan jumlah uang yang dibayar (Rp): "))
jumlah_orang = int(input("Masukkan jumlah orang untuk iuran: "))

total_belanja = jumlah_buku * harga_satuan
diskon = total_belanja * 0.10
total_bayar = total_belanja - diskon
kembalian = uang_bayar - total_bayar
iuran_per_orang = total_bayar // jumlah_orang
sisa_uang = total_bayar % jumlah_orang
uang_cukup = uang_bayar >= total_bayar

print(f"Total\t\t: {int(total_belanja)}")
print(f"Diskon\t\t: {int(diskon)}")
print(f"Bayar\t\t: {int(total_bayar)}")
print(f"Kembalian\t: {int(kembalian)}")
print(f"Per orang\t: {int(iuran_per_orang)}")
print(f"Sisa\t\t: {int(sisa_uang)}")
print(f"Uang cukup?\t: {uang_cukup}")
```

```python
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel_wp = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan rata-rata penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (misal 0.8 untuk 80%): "))
kebutuhan_rumah = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))

energi_harian = (jumlah_panel * (daya_panel_wp / 1000) * jam_penyinaran) * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik
kekurangan = kebutuhan_rumah - energi_harian
mencukupi = energi_harian >= kebutuhan_rumah

print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
daya_generator_kw = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_solar_per_kwh = float(input("Masukkan konsumsi bahan bakar (liter per kWh): "))
harga_solar_per_liter = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus_kabel = float(input("Masukkan arus pada kabel (A): "))
hambatan_kabel = float(input("Masukkan hambatan kabel (Ohm): "))

energi_harian = daya_generator_kw * jam_operasi
solar_per_hari = energi_harian * konsumsi_solar_per_kwh
biaya_solar = solar_per_hari * harga_solar_per_liter
tangki_cukup_hari = kapasitas_tangki / solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari
rugi_kabel = (arus_kabel ** 2) * hambatan_kabel
rugi_tinggi = (rugi_kabel > 150) and (arus_kabel >= 20)

print(f"Energi (kWh)\t: {int(energi_harian)}")
print(f"Solar (liter)\t: {solar_per_hari}")
print(f"Biaya (Rp)\t: {biaya_solar}")
print(f"Tangki cukup\t: {tangki_cukup_hari} hari")
print(f"Sisa solar (L)\t: {sisa_solar}")
print(f"Rugi kabel (W)\t: {rugi_kabel}")
print(f"Rugi tinggi?\t: {rugi_tinggi}")
```
