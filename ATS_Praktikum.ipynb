```python
#Nama:Danu Marsya
#NIm:4232501041
#tgl:10/11/2026
#Matkul:Pemerograman Dasar
#Soal 8

# Inisialisasi volume awal tangki
volume = 100

# Perulangan while selama volume masih di atas 0
while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")

    # Periksa jika volume mencapai tepat 20 liter
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah loop selesai
print("TANGKI KOSONG - Pengosongan berhenti.")
```

```python
#Nama:Danu Marsya
#NIm:4232501041
#tgl:10/11/2026
#Matkul:Pemerograman Dasar
#Soal 10

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

# Melakukan iterasi pada data dengan nomor titik dimulai dari 1
for i, nilai_str in enumerate(data, 1):
    try:
        # a. Ubah data menjadi float
        nilai = float(nilai_str)
        jumlah_valid += 1  # Hitung data valid yang berhasil diperiksa

        # b. Periksa jika nilai melebihi 70.0
        if nilai > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        # Jika gagal (misalnya "N/A"), cetak pesan dan hentikan perulangan
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

# c. Tampilkan jumlah data valid setelah perulangan selesai
print(f"Jumlah data valid: {jumlah_valid}")
```

```python
#Nama:Danu Marsya
#NIm:4232501041
#tgl:10/11/2026
#Matkul:Pemerograman Dasar
#Bonus Nilai

# 1. Menggunakan List untuk menyimpan data pembacaan beban daya (dalam kW)
# Terdapat campuran data angka dan satu data teks "ERROR" untuk menguji try-except.
data_beban_harian = [320.5, 415.0, 490.2, "ERROR", 560.0, 310.2]

# Definisi batas ambang (threshold) daya gardu
BATAS_SIAGA = 400.0
BATAS_OVERLOAD = 500.0

print("=== LAPORAN PEMANTAUAN BEBAN HARIAN GARDU LISTRIK ===")

# Inisialisasi variabel penghitung statistik
total_data_valid = 0
jumlah_overload = 0

# 2. Menggunakan perulangan (for loop) untuk memproses setiap data di dalam list
# Fungsi enumerate digunakan untuk melacak urutan jam/pembacaan mulai dari angka 1
for jam, item_beban in enumerate(data_beban_harian, start=1):

    # 3. Menggunakan try-except untuk mencegah program crash akibat data rusak
    try:
        # Mencoba mengonversi elemen list menjadi tipe data desimal (float)
        beban_kw = float(item_beban)
        total_data_valid += 1

        # 4. Menggunakan percabangan (if-elif-else) untuk menentukan status beban
        if beban_kw >= BATAS_OVERLOAD:
            status = "BAHAYA (OVERLOAD)"
            jumlah_overload += 1
        elif beban_kw > BATAS_SIAGA:
            status = "SIAGA (WASPADA)"
        else:
            status = "NORMAL"

        # Menampilkan hasil pemantauan per titik waktu
        print(f"Jam Ke-{jam}: {beban_kw} kW -> Status: {status}")

    except ValueError:
        # Blok ini akan dieksekusi jika data tidak bisa diubah ke float (misal: "ERROR")
        print(f"Jam Ke-{jam}: Data sensor '{item_beban}' tidak valid - Diabaikan.")

# Menampilkan ringkasan akhir setelah perulangan selesai
print("-" * 52)
print(f"Ringkasan Operasional:")
print(f"- Total data valid yang berhasil diperiksa: {total_data_valid}")
print(f"- Peringatan Overload terdeteksi: {jumlah_overload} kali")
```
