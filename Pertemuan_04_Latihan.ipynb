```python
#jawaban 1

tahun = int(input("Masukkan tahun: "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
print(f"Tahun {tahun} adalah tahun kabisat? {kabisat}")
```

```python
#jawaban 2

harga_per_cup = int(input("Masukkan harga kopi per cup: "))
jumlah_cup = int(input("Masukkan jumlah cup yang ingin dibeli: "))
uang_bayar = int(input("Masukkan jumlah uang yang dibayar: "))

total_harga = harga_per_cup * jumlah_cup
kembalian = uang_bayar - total_harga
uang_kurang = uang_bayar < total_harga

print(f"Total harga yang harus dibayar: Rp {total_harga}")
print(f"Kembalian transaksi: Rp {kembalian}")
print(f"Uangnya kurang! {uang_kurang}")
```
